```python

stock = {'LED': 20, 'RESISTOR': 50, 'SENSOR': 4}
requested = ['LED', 'SENSOR', 'SWITCH']

for item in requested:
    if item in stock:
        print(item, ":", stock[item], "available")
    else:
        print(item, ": unavailable")
```

```python

A = {101, 102, 103, 105}
B = {102, 104, 105, 106}

print("Both courses:", sorted(A & B))
print("Only Course A:", sorted(A - B))
print("Either course:", sorted(A | B))
```

```python

marks = {'Arun': 85, 'Priya': 92, 'Kavin': 78}

result = {}

for name, mark in marks.items():
    result[mark] = name

print(result)
```

```python

names = ['Anu', 'Arun', 'Bala', 'Banu', 'Cathy', 'Asha']

result = {}

for name in names:
    key = name[0].upper()

    if key not in result:
        result[key] = []

    result[key].append(name)

print(result)
```

```python

store_a = {'LED': 20, 'Cable': 10, 'Board': 3}
store_b = {'Cable': 5, 'Sensor': 7, 'LED': 8}

result = store_a.copy()

for item in store_b:
    if item in result:
        result[item] += store_b[item]
    else:
        result[item] = store_b[item]

print("Combined inventory:", result)
```

```python

sales = {'Sensor': [4, 6, 5],
         'Cable': [10, 2, 3],
         'Board': [8, 7],
         'LED': [5, 5, 5]}

result = []

for item in sales:
    total = sum(sales[item])
    result.append((item, total))

result.sort(key=lambda x: (-x[1], x[0]))

for item, total in result:
    print(item, ":", total)
```

```python

records = [
    ('D01', 'Lab-A', 'READY'),
    ('D02', 'Lab-B', 'BUSY'),
    ('D01', 'Lab-C', 'MAINTENANCE')
]

result = {}

for id, location, status in records:
    result[id] = {'location': location, 'status': status}

print(result)
```

```python

permissions = {
    'admin': {'read', 'write', 'delete'},
    'operator': {'read', 'write'},
    'auditor': {'read', 'export'}
}

common = set.intersection(*permissions.values())
all_permissions = set.union(*permissions.values())

print("Common permissions:", sorted(common))
print("All permissions:", sorted(all_permissions))
```

```python

nums = [1, 1, 1]
k = 2

count = 0
prefix = 0
d = {0: 1}

for n in nums:
    prefix += n
    count += d.get(prefix - k, 0)
    d[prefix] = d.get(prefix, 0) + 1

print("Count:", count)
```

```python

s = "ADOBECODEBANC"
t = "ABC"

result = ""

for i in range(len(s)):
    for j in range(i + 1, len(s) + 1):
        sub = s[i:j]
        if all(sub.count(c) >= t.count(c) for c in t):
            if result == "" or len(sub) < len(result):
                result = sub
            break

print("Minimum window:", result)
```
