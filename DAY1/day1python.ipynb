```python

c = float(input("Enter Celsius: "))

f = (c * 9/5) + 32
k = c + 273.15

print(f"Celsius: {c:.2f}°C")
print(f"Fahrenheit: {f:.2f}°F")
print(f"Kelvin: {k:.2f} K")

print("\nCelsius   Fahrenheit   Kelvin")
for c in range(-40, 101, 10):
    f = (c * 9/5) + 32
    k = c + 273.15
    print(f"{c:7.2f} {f:12.2f} {k:10.2f}")
```

```python

a, b, c = map(int, input("Enter three numbers: ").split())

print("Largest:", max(a, b, c))
print("Smallest:", min(a, b, c))
print("Average:", round((a + b + c) / 3, 2))

n = int(input("Enter a number: "))

if n > 0:
    if n % 2 == 0:
        print("Classification: Positive and Even")
    else:
        print("Classification: Positive and Odd")
elif n < 0:
    if n % 2 == 0:
        print("Classification: Negative and Even")
    else:
        print("Classification: Negative and Odd")
else:
    print("Classification: Zero")
```

```python

year = int(input("Enter year: "))

if year % 400 == 0 or (year % 4 == 0 and year % 100 != 0):
    print("Leap Year")
else:
    print("Not a Leap Year")

marks = int(input("Enter marks: "))

if marks < 0 or marks > 100:
    print("Invalid marks")
elif marks >= 90:
    print("Grade A+")
elif marks >= 75:
    print("Grade A")
elif marks >= 50:
    print("Grade B")
else:
    print("Fail")
```

```python

total = 0

while True:
    s = input("Enter price (done to stop): ")

    if s == "done":
        break

    total += float(s)

print("Subtotal:", total)
```

```python

a = int(input("Enter a: "))
b = int(input("Enter b: "))

print("Before Swap:", a, b)

a, b = b, a

print("After Swap:", a, b)
```

```python

p = float(input("Enter principal: "))
r = float(input("Enter rate: "))
t = float(input("Enter time: "))

if p >= 0 and t >= 0:
    si = (p * r * t) / 100
    print("Simple Interest:", si)
    print("Total Amount:", p + si)
else:
    print("Invalid input")
```

```python

x = input("Enter a value: ")

print("Original type:", type(x))

try:
    print("Integer:", int(x))
    print("Integer type:", type(int(x)))
except ValueError:
    print("Invalid integer conversion")

try:
    print("Float:", float(x))
    print("Float type:", type(float(x)))
except ValueError:
    print("Invalid float conversion")
```
