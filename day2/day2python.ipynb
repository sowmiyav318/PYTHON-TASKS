```python

n = int(input("Enter N: "))
num = int(input("Enter number: "))

print("Prime numbers:")
for i in range(2, n + 1):
    for j in range(2, i):
        if i % j == 0:
            break
    else:
        print(i, end=" ")

fact = 1
for i in range(1, n + 1):
    fact *= i
print("\nFactorial:", fact)

a, b = 0, 1
print("Fibonacci:", end=" ")
for i in range(n):
    print(a, end=" ")
    a, b = b, a + b

s = 0
rev = 0
temp = num
while temp > 0:
    digit = temp % 10
    s += digit
    rev = rev * 10 + digit
    temp //= 10

print("\nDigit Sum:", s)
print("Reverse:", rev)
```

```python

n = int(input("Enter N: "))

print("Right Triangle")
for i in range(1, n + 1):
    print("*" * i)

print("Number Triangle")
for i in range(1, n + 1):
    for j in range(1, i + 1):
        print(j, end="")
    print()

print("Multiplication Table")
for i in range(1, 11):
    for j in range(1, 11):
        print(i * j, end="\t")
    print()
```

```python

s = input("Enter a string: ")

print("Reverse:", s[::-1])

if s.replace(" ", "").lower() == s.replace(" ", "").lower()[::-1]:
    print("Palindrome: Yes")
else:
    print("Palindrome: No")

v = c = d = 0

for ch in s.lower():
    if ch in "aeiou":
        v += 1
    elif ch.isalpha():
        c += 1
    elif ch.isdigit():
        d += 1

print("Vowels:", v)
print("Consonants:", c)
print("Digits:", d)
```

```python

name = input("Enter customer name: ")
units = int(input("Enter units: "))

if units <= 100:
    bill = units * 3
elif units <= 200:
    bill = 300 + (units - 100) * 4
else:
    bill = 700 + (units - 200) * 5

print("Customer:", name)
print("Units:", units)
print("Amount: Rs.", format(bill, ".2f"))
```

```python

amount = int(input("Enter withdrawal amount: "))

if amount > 20000 or amount <= 0 or amount % 10 != 0:
    print("Invalid amount")
else:
    for note in [500, 200, 100, 50, 10]:
        count = amount // note
        print(note, "notes:", count)
        amount %= note
```

```python

entry = input("Enter entry time (HH:MM): ")
exit = input("Enter exit time (HH:MM): ")

h1, m1 = map(int, entry.split(":"))
h2, m2 = map(int, exit.split(":"))

start = h1 * 60 + m1
end = h2 * 60 + m2

if end < start:
    end += 1440

duration = end - start
hours = duration // 60
minutes = duration % 60

bill_hours = (duration + 59) // 60

if bill_hours == 0:
    fee = 0
else:
    fee = 30 + (bill_hours - 1) * 20

print("Parking Duration:", hours, "hours", minutes, "minutes")
print("Billable Hours:", bill_hours)
print("Parking Fee: Rs.", fee)
```

```python

n = int(input("Enter number of deliveries: "))
total_distance = 0
earnings = 0

for i in range(n):
    d = float(input("Enter distance: "))
    total_distance += d

    if d <= 5:
        earnings += 40
    else:
        earnings += 40 + (d - 5) * 8

print("Total Distance:", round(total_distance, 2), "km")
print("Total Earnings: Rs.", round(earnings, 2))
print("Average Distance:", round(total_distance / n, 2), "km")
```

```python

x = int(input("Enter a number: "))
rev = 0
n = abs(x)

while n > 0:
    digit = n % 10
    rev = rev * 10 + digit
    n = n // 10

if x < 0:
    rev = -rev

print("Reverse:", rev)
```
