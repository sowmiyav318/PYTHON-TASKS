```python

n = int(input("How many numbers? "))
a = list(map(int, input("Enter numbers: ").split()))

positive = negative = zero = 0
even = odd = 0
both = only3 = only5 = neither = 0

for x in a:
    if x > 0:
        positive += 1
    elif x < 0:
        negative += 1
    else:
        zero += 1

    if x != 0:
        if x % 2 == 0:
            even += 1
        else:
            odd += 1

    if x % 3 == 0 and x % 5 == 0:
        both += 1
    elif x % 3 == 0:
        only3 += 1
    elif x % 5 == 0:
        only5 += 1
    else:
        neither += 1

print("Positive:", positive)
print("Negative:", negative)
print("Zero:", zero)
print("Even:", even)
print("Odd:", odd)
print("Divisible by both 3 and 5:", both)
print("Divisible only by 3:", only3)
print("Divisible only by 5:", only5)
print("Divisible by neither:", neither)
```

```python

count = 0

while True:
    n = int(input("Enter a number (1-100): "))

    if n < 1 or n > 100:
        print("Invalid input")
        count += 1
        continue

    print("Accepted value:", n)
    print("Invalid attempts:", count)
    break
```

```python

while True:
    print("1. Celsius to Fahrenheit")
    print("2. Kilometres to Miles")
    print("3. Kilograms to Pounds")
    print("4. Exit")

    choice = int(input("Enter choice: "))

    if choice == 1:
        c = float(input("Enter Celsius: "))
        print("Fahrenheit:", (c * 9/5) + 32)

    elif choice == 2:
        km = float(input("Enter kilometres: "))
        print("Miles:", round(km * 0.621371, 2))

    elif choice == 3:
        kg = float(input("Enter kilograms: "))
        print("Pounds:", round(kg * 2.20462, 2))

    elif choice == 4:
        print("Exit")
        break

    else:
        print("Invalid choice")
        continue
```

```python

n = int(input("Enter a number: "))
count = 0

for i in range(1, 21):
    result = n * i

    if result > 100:
        break

    if result % 3 == 0:
        count += 1
        continue

    print(n, "x", i, "=", result)

print("Skipped results:", count)
```

```python

s = input("Enter password: ")

upper = lower = digit = special = False

for ch in s:
    if ch.isupper():
        upper = True
    elif ch.islower():
        lower = True
    elif ch.isdigit():
        digit = True
    else:
        special = True

if len(s) >= 8 and upper and lower and digit and special:
    print("Valid password")
else:
    print("Invalid password")

print("Uppercase:", upper)
print("Lowercase:", lower)
print("Digit:", digit)
print("Special character:", special)
print("Minimum length:", len(s) >= 8)
```

```python

n = int(input("Enter size: "))

for i in range(n):
    for j in range(n):
        if i == 0 or i == n-1 or j == 0 or j == n-1:
            print("*", end="")
        else:
            print(" ", end="")
    print()
```

```python

l = int(input("Enter L: "))
r = int(input("Enter R: "))

found = False

for i in range(l, r + 1):
    if i >= 0:
        j = 0
        while j * j < i:
            j += 1

        if j * j == i:
            print("First perfect square:", i)
            found = True
            break

if not found:
    print("No perfect square found")
```

```python

secret = 8

for i in range(5):
    guess = int(input("Guess the number: "))

    if guess == secret:
        print("Correct")
        break
    elif guess < secret:
        print("Too Low")
    else:
        print("Too High")
else:
    print("Game Over")
```

```python

n = int(input("Enter count: "))
a = list(map(int, input("Enter numbers: ").split()))

best = 1
count = 1
value = a[0]
current = a[0]

for i in range(1, n):
    if a[i] == a[i-1]:
        count += 1
    else:
        count = 1

    if count > best:
        best = count
        value = a[i]

print("Value =", value)
print("Length =", best)
```

```python

normal = large = suspicious = 0
total = 0

while True:
    n = int(input("Enter amount (-1 to stop): "))

    if n == -1:
        break

    if n == 0:
        pass
    elif n < 0:
        continue
    elif n <= 1000:
        normal += 1
        total += n
    elif n <= 5000:
        large += 1
        total += n
    else:
        suspicious += 1
        total += n

        if suspicious == 3:
            break

print("Normal transactions:", normal)
print("Large transactions:", large)
print("Suspicious transactions:", suspicious)
print("Total valid amount:", total)
```
