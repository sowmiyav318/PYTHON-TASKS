```python

try:
    a = int(input("Integer: "))
    b = float(input("Decimal: "))

    print("Integer Value:", a)
    print("Integer Type:", type(a).__name__)
    print("Float Value:", b)
    print("Float Type:", type(b).__name__)
    print("Rounded Value:", round(b))

    if b != 0:
        print("Division:", a / b)
        print("Floor Division:", a // b)
        print("Modulus:", a % b)

    print("Power:", a ** 2)

except ValueError:
    print("Invalid input")
```

```python

msg = input("Message: ")
shift = int(input("Shift: "))

encrypted = ""

for ch in msg:
    if ch.isupper():
        encrypted += chr((ord(ch) - 65 + shift) % 26 + 65)
    elif ch.islower():
        encrypted += chr((ord(ch) - 97 + shift) % 26 + 97)
    else:
        encrypted += ch

print("Encrypted:", encrypted)

decrypted = ""

for ch in encrypted:
    if ch.isupper():
        decrypted += chr((ord(ch) - 65 - shift) % 26 + 65)
    elif ch.islower():
        decrypted += chr((ord(ch) - 97 - shift) % 26 + 97)
    else:
        decrypted += ch

print("Decrypted:", decrypted)
```

```python

start = int(input("Start: "))
end = int(input("End: "))

for n in range(start, end + 1):
    s = str(n)
    digit_sum = sum(int(d) for d in s)
    digits = len(s)
    binary = bin(n)[2:]

    prime = n > 1
    for i in range(2, n):
        if n % i == 0:
            prime = False
            break

    total = 0
    for i in range(1, n):
        if n % i == 0:
            total += i
    perfect = n > 0 and total == n

    armstrong = sum(int(d) ** len(s) for d in s) == n
    palindrome = s == s[::-1]

    print("\nNumber:", n)
    print("Prime:", prime)
    print("Perfect:", perfect)
    print("Armstrong:", armstrong)
    print("Palindrome:", palindrome)
    print("Digit Sum:", digit_sum)
    print("Digits:", digits)
    print("Binary:", binary)
```

```python

a = int(input("Number 1: "))
b = int(input("Number 2: "))

n = a
binary = ""
while n > 0:
    binary = str(n % 2) + binary
    n = n // 2

n = a
octal = ""
while n > 0:
    octal = str(n % 8) + octal
    n = n // 8

n = a
digits = "0123456789ABCDEF"
hexa = ""
while n > 0:
    hexa = digits[n % 16] + hexa
    n = n // 16

x, y = a, b
while y != 0:
    x, y = y, x % y

gcd = x
lcm = (a * b) // gcd

print("Binary:", binary)
print("Octal:", octal)
print("Hexadecimal:", hexa)
print("GCD:", gcd)
print("LCM:", lcm)
```

```python

n = int(input("N: "))
d1 = int(input("Divisor 1: "))
w1 = input("Word 1: ")
d2 = int(input("Divisor 2: "))
w2 = input("Word 2: ")

for i in range(1, n + 1):
    if i % d1 == 0 and i % d2 == 0:
        print(w1 + w2)
    elif i % d1 == 0:
        print(w1)
    elif i % d2 == 0:
        print(w2)
    else:
        print(i)
```

```python

n = int(input("Students: "))

for i in range(n):
    name = input("Student name: ")
    marks = input("Enter 3 marks: ").split()

    try:
        marks = [int(m) for m in marks]

        if len(marks) != 3 or any(m < 0 or m > 100 for m in marks):
            print(name, "Invalid marks")
        else:
            total = sum(marks)
            avg = total / 3

            if avg >= 90:
                grade = "A+"
            elif avg >= 75:
                grade = "A"
            elif avg >= 50:
                grade = "B"
            else:
                grade = "Fail"

            print("Student:", name)
            print("Total:", total)
            print("Average:", round(avg, 2))
            print("Grade:", grade)

    except ValueError:
        print(name, "Invalid marks")
```

```python

n = int(input("Transactions: "))
amounts = []
rejected = 0

for i in range(n):
    s = input("Enter amount: ")

    try:
        amount = float(s)

        if amount < 0:
            rejected += 1
        else:
            amounts.append(amount)

    except ValueError:
        rejected += 1

if len(amounts) > 0:
    print("Successful Transactions:", len(amounts))
    print("Total Amount: ₹", round(sum(amounts), 2))
    print("Highest Transaction: ₹", max(amounts))
    print("Lowest Transaction: ₹", min(amounts))
    print("Average Transaction: ₹", round(sum(amounts) / len(amounts), 2))

print("Rejected Transactions:", rejected)
```

```python

s = input("Enter a string: ")
s = s.lstrip()
num = 0
sign = 1
i = 0

if s and s[0] == '-':
    sign = -1
    i = 1
elif s and s[0] == '+':
    i = 1

while i < len(s) and s[i].isdigit():
    num = num * 10 + int(s[i])
    i += 1

print(sign * num)
```

```python

a = input("Enter first binary: ")
b = input("Enter second binary: ")

n1 = int(a, 2)
n2 = int(b, 2)

total = n1 + n2
result = ""

while total > 0:
    result = str(total % 2) + result
    total = total // 2

print("Sum:", result if result else "0")
```
