```python

def product_except_self(nums):
    n = len(nums)
    result = [1] * n

    prefix = 1
    for i in range(n):
        result[i] = prefix
        prefix *= nums[i]

    suffix = 1
    for i in range(n - 1, -1, -1):
        result[i] *= suffix
        suffix *= nums[i]

    return result


nums = [1, 2, 3, 4]

print("Product array:", product_except_self(nums))
```

```python

def calculate_bill(prices, tax_rate=0.10, discount=0.20):
    subtotal = sum(prices)
    discount_amount = subtotal * discount
    taxable_amount = subtotal - discount_amount
    tax = taxable_amount * tax_rate
    total = taxable_amount + tax

    print("Subtotal:", format(subtotal, ".2f"))
    print("Discount:", format(discount_amount, ".2f"))
    print("Tax:", format(tax, ".2f"))
    print("Final total:", format(total, ".2f"))


prices = [100, 50]

calculate_bill(prices, tax_rate=0.10, discount=0.20)
```

```python

def average_readings(*args):
    if len(args) == 0:
        return None

    return sum(args) / len(args)


result = average_readings(10, 20, 30, 40)

if result is not None:
    print("Average:", result)
else:
    print("No readings provided")


result = average_readings()

if result is not None:
    print("Average:", result)
else:
    print("No readings provided")
```

```python

def create_device_config(device_id, **kwargs):
    accepted_fields = ["baud_rate", "parity", "timeout_ms", "debug"]

    config = {"device_id": device_id}

    for key, value in kwargs.items():
        if key in accepted_fields:
            config[key] = value
        else:
            print("Unsupported key:", key)

    return config


result = create_device_config(
    "D01",
    baud_rate=115200,
    parity="N",
    timeout_ms=500
)

print(result)
```

```python

def group_anagrams(strs):
    groups = {}

    for word in strs:
        key = ''.join(sorted(word))

        if key not in groups:
            groups[key] = []

        groups[key].append(word)

    return list(groups.values())


strs = ["eat", "tea", "tan", "ate", "nat", "bat"]

result = group_anagrams(strs)

print("Groups:")
for group in result:
    print(group)
```

```python

def top_k_frequent(nums, k):
    frequency = {}

    for num in nums:
        if num in frequency:
            frequency[num] += 1
        else:
            frequency[num] = 1

    sorted_nums = sorted(frequency, key=frequency.get, reverse=True)

    return sorted_nums[:k]


nums = [1, 1, 1, 2, 2, 3]
k = 2

result = top_k_frequent(nums, k)

print("Top", k, "frequent values:", result)
```

```python

def longest_consecutive(nums):
    num_set = set(nums)
    longest = 0
    best_sequence = []

    for num in num_set:
        if num - 1 not in num_set:
            current = num
            count = 1
            sequence = [num]

            while current + 1 in num_set:
                current += 1
                count += 1
                sequence.append(current)

            if count > longest:
                longest = count
                best_sequence = sequence

    print("Longest consecutive sequence length:", longest)
    print("Sequence:", ", ".join(map(str, best_sequence)))


nums = [100, 4, 200, 1, 3, 2]

longest_consecutive(nums)
```

```python

def find_max_length(nums):
    balance = 0
    max_length = 0
    balance_map = {0: -1}

    for i in range(len(nums)):
        if nums[i] == 1:
            balance += 1
        else:
            balance -= 1

        if balance in balance_map:
            length = i - balance_map[balance]
            max_length = max(max_length, length)
        else:
            balance_map[balance] = i

    return max_length


nums = [0, 1, 0]

result = find_max_length(nums)

print("Maximum balanced subarray length:", result)
```

```python

def length_of_last_word(s):
    words = s.split()
    return len(words[-1])


s = " Embedded Python "

result = length_of_last_word(s)

print("Length of last word:", result)
```

```python

def str_str(haystack, needle):
    if needle == "":
        return 0

    for i in range(len(haystack) - len(needle) + 1):
        if haystack[i:i + len(needle)] == needle:
            return i

    return -1


haystack = "datascience"
needle = "science"

result = str_str(haystack, needle)

print("First occurrence index:", result)
```
